# Python Example for Linear Regression

Let's put the concepts we have discussed into practice with a simple linear regression model. We will use a small dataset where the goal is to predict exam scores based on the number of hours a student studies.

## Step 1: Import the Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

## Step 2: Create the Data

```python
# Hours studied
X = np.array([[1], [2], [3], [4], [5], [6], [7], [8]])

# Exam scores
y = np.array([52, 55, 61, 65, 70, 74, 78, 85])
```

Here, X represents our input feature, while y represents our target variable.

## Step 3: Train the Model

```python
model = LinearRegression()
model.fit(X, y)
```

The fit() method trains the model by finding the intercept and coefficient that provide the best fit to the training data.

## Step 4: Make Predictions

```python
predictions = model.predict(X)

print(predictions)
```

The model uses the learned regression equation to predict an exam score for each number of study hours.

We can also predict the score for a new student who studies for 9 hours:

```python
new_prediction = model.predict([[9]])

print(new_prediction)
```

## Step 5: Examine the Coefficients

```python
print("Coefficient:", model.coef_[0])
print("Intercept:", model.intercept_)
```

The coefficient tells us how much the predicted exam score changes for each additional hour of studying, while the intercept represents the model's predicted score when the number of study hours is zero.

In our example, the linear regression model learns the following equation:

$$ ŷ = 46.61 + 4.64x $$

Here, **46.61** is the **intercept**, while **4.64** is the **coefficient**. This means that, according to the model, each additional hour of studying is associated with an estimated **4.64-point increase** in exam score.


## Step 6: Visualize the Regression Line

```python
plt.scatter(X, y, label="Actual Data")
plt.plot(X, predictions, label="Regression Line")

plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.title("Linear Regression: Study Hours vs. Exam Score")
plt.legend()
plt.show()
```

The scatter plot shows the actual observations, while the line represents the model's predictions.

This simple example demonstrates the complete idea behind linear regression: the model learns a line from the training data and then uses that line to predict continuous numerical values.

## Visualizing Residuals

```python
plt.figure(figsize=(10, 6))

plt.scatter(X, y, s=70, label="Actual Data")
plt.plot(X, predictions, linewidth=2, label="Regression Line")

for i in range(len(X)):
    plt.plot(
        [X[i], X[i]],
        [y[i], predictions[i]],
        linestyle="--",
        linewidth=2,
        color="red"
    )

plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.title("Linear Regression and Residuals")
plt.legend()

plt.show()
```

The graph above illustrates how linear regression fits a line to the observed data. The dots represent the actual exam scores, while the line represents the predicted scores generated by the model. The red dashed vertical lines show the residuals, or the differences between the actual and predicted values. A shorter residual means the prediction is closer to the actual value, while a longer residual indicates a larger error. The model aims to find the regression line that minimizes these errors across all observations.

```python
# Calculate residuals
residuals = y - predictions

plt.figure(figsize=(8, 5))

# Plot residuals
plt.scatter(X, residuals, s=70)

# Add a zero-error reference line
plt.axhline(y=0, linestyle="--")

plt.xlabel("Hours Studied")
plt.ylabel("Residual (Actual − Predicted)")
plt.title("Residuals in Linear Regression")

plt.show()
```

* Residual = 0 → prediction is exactly correct
* Positive residual → model underestimated the actual value
* Negative residual → model overestimated the actual value
* Larger distance from 0 → larger prediction error
